# 📌 Step 2: Data Cleaning

## We will check and clean:

✔ Drop duplicates

✔ Handle missing values

✔ Fix wrong data types

✔ Clean “rate” column

## Example problems:

Some ratings are "NEW", "nan", "-"

Votes/rating sometimes stored as text

Cuisines, location, etc. may need cleaning

```python
# Clean Rating Column
df['Aggregate rating'] = df['Aggregate rating'].astype(str) 

# Whatever values the rating columns has, whether its a number, text or string, 
# convert everything into string so that we can clean them easily.

df['Aggregate rating'] = df['Aggregate rating'].apply(
    lambda x: x.split('/')[0] if '/' in x else x
) 

# Some datasets store ratings like "4.5/5" or "3.5/5". This line checks if "/" is present.
# If present, it splits "4.5/5" into ["4.5","5"].
# If not ("NEW" or "nan"), it remains the same.

df['Aggregate rating'] = df['Aggregate rating'].replace(['NEW', '-', 'nan'], np.nan) 

# Some entries like "NEW","-" and "nan" are not actual ratings(not valid numerical values), so replace them with NaN (missing value).

df['Aggregate rating'] = pd.to_numeric(df['Aggregate rating'], errors='coerce')

# Converts the cleaned string values ("4.5") into actual numerical floats (4.5).
# If by mistake any string cannot be converted, errors="coerce" turns it into NaN instead of throwing an error.
```

```python
# Clean yes/No Columns
df['Has Online delivery'] = df['Has Online delivery'].map({'Yes': 1, 'No': 0})
df['Has Table booking'] = df['Has Table booking'].map({'Yes': 1, 'No': 0})
```

```python
# Clean String columns

# str.strip() removes extra spaces at the beginning of the text and at the end.
# Improves consistency and duplicates created by extra spaces.
# Ensures proper grouping when value_counts() or sorting is done.

df['City'] = df['City'].str.strip()
df['Cuisines'] = df['Cuisines'].str.strip()
df['Restaurant Name'] = df['Restaurant Name'].str.strip()
```

```python
# Check how many times these appear.
# value_counts() - a dictionary-like series.
# .map() - replaces each city with its count.

df['City_count'] = df['City'].map(df['City'].value_counts())

# Checks how many times each restaurant appears.
# Helps detect duplicate restaurants or chains.

df['Restaurant_count'] = df['Restaurant Name'].map(df['Restaurant Name'].value_counts())

# Check how many times each cuisine combination appears.
# Finds how many restaurants serve the same cuisine combination.

df['Cuisine_count'] = df['Cuisines'].map(df['Cuisines'].value_counts())
```
