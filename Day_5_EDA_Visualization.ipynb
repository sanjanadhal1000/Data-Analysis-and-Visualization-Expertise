# Is the rating distribution skewed or symmetric?

# What does the spread tell you?

Answer: Your histogram of Aggregate Rating shows:

A huge spike at 0 (around 2000+ values).

The rest of the ratings cluster mostly between 2.5 and 4.5, with fewer observations above 4.5.

This means:

👉 The distribution is highly right-skewed (positively skewed).

Why?
Because a large number of restaurants have a rating of 0, which pulls the distribution heavily toward the left side, creating a long tail to the right.

Without the 0's, the remaining distribution (2–5) would be much more symmetrical, but overall it is not symmetric due to the zero spike.

## Which column has more extreme outliers: Votes or Cost?
- 'Cost' has more extreme outliers. 
- There are values going from 0 to 800000+.
- The main distribution is squeezed at the bottom (99% of the values are small, but a few restaurants have massive prices).
- These are extreme outliers - orders of high magnitude.

## What might be causing those outliers?
## Data entry errors (Most Common): 
Someone may have typed 800000 instead of 800, for example or 200000 instead of 2000. These are classic mistakes in price fields.

## Currency Conversion Inconsistencies:
Zomato dataset includes restaurants from many countries.
If the price is stored in USD/EUR and not converted to INR values, then Rs.2000 vs $2000. $2000 = 1.6 lakh rupees.
Huge jumps will appear.

## Premium Luxury Restaurants
Some luxury restaurants may cost Rs. 20,000 to 50,000 for two, which is rare but possible. But this is unrealistic, causing errors.

## Missing Value Encoding
Sometimes 0 or extremely high values are used as placeholders for missing data.

    1. Votes ↔ Aggregate Rating

    So: weak-to-moderate positive correlation.
    
    2. Average Cost for Two ↔ Aggregate Rating
    
    The points look very scattered.
    
    No clear upward or downward trend.
    
    So: no visible correlation.
    
    3. Price Range ↔ Aggregate Rating
    
    The price range is categorical (1–4), and ratings are spread across all of them.
    
    No visible trend.
    
    So: no correlation.
    
    4. Price Range ↔ Average Cost for Two
    
    This should correlate because price range corresponds to cost, but in your plot the points are very scattered vertically.
    
    You might be seeing:
    
    Wrong scaling,
    
    Outliers,
    
    Or messy data.
    
    So: expected correlation is not clearly visible due to noise.

## Are there visible clusters?
